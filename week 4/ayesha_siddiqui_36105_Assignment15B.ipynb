```python
import  pandas as pd
```

```python
df = pd.read_csv("student_tea_snacks.csv")
```

```python
df["tea_type"].value_counts()
# the most preferred tea type
```

```python
df.groupby('gender')["cups_per_day"].agg(['mean', 'sum', 'count'])
#average number of cups per gender
```

```python
import matplotlib.pyplot as plt
snack_counts = df["snack_choice"].value_counts()
plt.bar(snack_counts.index, snack_counts.values, color="skyblue", width=0.5)
plt.xlabel("Snack Choice")
plt.ylabel("Number of Students")
plt.title("Snack Choices")
plt.show()
#a bar chart of snack choices
```

```python
compiled = df.groupby('age_group')['tea_type'].value_counts()
compiled
#Comparing tea preference by age group
```

```python
no_tea = df['tea_type'].isna().sum()
no_tea
#counting the number of students who don't drink tea
```
