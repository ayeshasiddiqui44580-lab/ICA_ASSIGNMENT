```python
import pandas as pd
```

```python
df = pd.read_csv("ride_hailing_dataset.csv")
```

```python
df.groupby("service_type")["fare"].agg(["count", "mean", "median", "std"])
#the average fare by service type
```

```python
df["city"].value_counts()
#Islamabad has the most trips
```

```python
df["distance_km"].mean()
#average distance per trip
```

```python
df.groupby("gender")["trip_rating"].agg(["count", "mean", "median", "std"])
#Comparing trip ratings by gender
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
avg_fare = df.groupby("service_type")["fare"].mean()
```

```python
plt.bar(avg_fare.index, avg_fare.values, color="gray", width=0.5)
plt.xlabel("Service Type")
plt.ylabel("Average Fare")
plt.title("Average Fare by Service Type")
plt.show()
```
