```python
#a relationship between screen time and sleep hours:
plt.scatter('screen_time', 'sleep_hours' , data=df2 , c='purple', s=50, alpha=1)
plt.xlabel('screen_time')
plt.ylabel('sleep_hours')
plt.title('Scatter Plot')
plt.show ()
#There is no correlation
```

```python
plt.hist(df2['daily_transport_cost'], bins=20, color='pink', edgecolor='black', alpha=0.7)

plt.xlabel('Daily Transport Cost')
plt.ylabel('Frequency')
plt.title('Histogram of Daily Transport Cost')
plt.show()
#a histogram of daily transport cost
```
